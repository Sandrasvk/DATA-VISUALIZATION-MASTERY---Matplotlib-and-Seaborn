```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

# Distribution plots

1. distplot
2. jointplot
3. pairplot
4. rugplot
5. kdeplot

```python
#distplot
sns.distplot(tips['total_bill'])
```

```python
sns.distplot(tips['total_bill'], kde = False)
```

```python
sns.distplot(tips['total_bill'], kde = True, hist = False)
```

# jointplot
scatter
reg
kde
hex

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = tips, kind ='scatter')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = tips, kind ='kde')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = tips, kind ='hex')
```

```python
#pairplot
sns.pairplot(tips)
```

```python
sns.pairplot(tips, diag_kind = 'kde', kind = 'scatter', hue = 'sex')
```

```python
sns.pairplot(tips, diag_kind = 'kde', kind = 'scatter', hue = 'sex', palette = 'husl')
```

```python
#rugplot
sns.rugplot(tips['total_bill'])
```

```python
#kdeplot
sns.kdeplot(tips['tip'])
```

# categorical plots

1. factorplot - catplot
2. boxplot
3. violinplot
4. stripplot
5. swarmplot
6. barplot
7. countplot

```python
#barplot/countplot
sns.barplot(x = 'sex', y = 'total_bill', data = tips)
```

```python
sns.countplot(x = 'sex', data = tips)
```

```python
#boxplot
sns.boxplot(x = 'day', y = 'total_bill', hue = 'smoker', data = tips, palette = 'rainbow')
```

```python
sns.boxplot(data = tips, palette = 'rainbow', orient = 'h')
```

```python
#violinplot
sns.violinplot(x = 'day', y = 'total_bill', data = tips, palette = 'rainbow')
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data = tips, hue = 'sex', split = True, palette = 'Set1')
```

```python
#stripplot
sns.stripplot(x = 'day', y = 'total_bill', data = tips, jitter = True)
```

```python
#swarmplot
sns.swarmplot(x = 'day', y = 'total_bill', hue = 'sex', data = tips, palette = 'Set1')
```

#factorplot
sns.factorplot(x = 'sex', y = 'total_bill', data = tips, kind = 'bar') #Factorplot only in old versions, new versions catplot instead of this

```python
#catplot
sns.catplot(x = 'sex', y = 'total_bill', data = tips, kind = 'bar')
```
