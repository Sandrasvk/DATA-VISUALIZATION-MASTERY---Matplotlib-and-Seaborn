# grids

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
iris['species'].unique()
```

```python
#pairgrid
import matplotlib.pyplot as plt
g = sns.PairGrid(iris)
plt.show()
```

```python
g.map(plt.scatter)
plt.show()
```

```python
g = sns.PairGrid(iris)
g.map(plt.bar)
plt.show()
```

```python
g = sns.PairGrid(iris)
g.map_diag(plt.hist)
g.map_upper(plt.scatter)
g.map_lower(sns.kdeplot)
plt.show()
```

```python
# facetGrid
g = sns.FacetGrid(tips, col = 'time', row = 'smoker')
g.map(sns.distplot, 'total_bill')
plt.show()
```

```python
g = sns.FacetGrid(tips, col = 'time', row = 'smoker', hue = 'sex')
g.map(plt.scatter, 'total_bill', 'tip').add_legend()
```

```python
#jointGrid
g = sns.JointGrid(x = 'total_bill', y = 'tip', data = tips)
plt.show()
```

```python
g.plot(sns.regplot, sns.distplot)
plt.show()
```

# Regression plots

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips)
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', palette = 'plasma', markers = ['o', 'v'], scatter_kws = {'s' : 100})
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, row = 'sex', col = 'time')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, col = 'day', hue = 'sex', palette = 'plasma', aspect = 1.5) #0.5, 1.0, 1.5
```

```python
# countplot
sns.countplot(x = 'sex', data = tips)
```

```python
import numpy as np
import matplotlib.pyplot as plt
def sinplot(flip = 1):
    x = np.linspace(0,14,100)
    for i in range(1,7):
        plt.plot(x, np.sin(x + 1 * .5) * (7 - i))
```

```python
sns.set_context('notebook')
sinplot()
```

```python
sns.set_context('paper')
sinplot()
plt.show()
```

```python
sns.set_context('talk')
sinplot()
plt.show()
```

```python
sns.set_context('poster')
sinplot()
plt.show()
```
